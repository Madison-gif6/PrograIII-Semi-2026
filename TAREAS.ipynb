```python
import math
from collections import Counter

# Dataset con 20 compañeros requeridos por la guía
estudiantes = [
    {"nombres": "Madison Odalis", "apellidos": "Valle Juárez", "edad": 20, "traslado": 30, "telefono": 8.5, "materias": {"Programación III": 10.0, "Base de Datos II": 9.0, "Inglés Básico": 10.0, "Diseño de Circuitos Digitales": 10.0, "Estadística Computacional": 9.0}},
    {"nombres": "Karen Elizabeth", "apellidos": "Ulloa Goméz", "edad": 18, "traslado": 40, "telefono": 9.0, "materias": {"Programación III": 10.0, "Base de Datos II": 10.0, "Inglés Básico": 10.0, "Diseño de Circuitos Digitales": 10.0, "Estadística Computacional": 10.0}},
    {"nombres": "Flor Zulema", "apellidos": "Campos Orellana", "edad": 20, "traslado": 6, "telefono": 5.38, "materias": {"Programación III": 9.0, "Base de Datos II": 9.0, "Inglés Básico": 9.0, "Diseño de Circuitos Digitales": 9.0, "Estadística Computacional": 9.0}},
    {"nombres": "Jimmy Arturo", "apellidos": "Marroquin Guzmán", "edad": 20, "traslado": 30, "telefono": 6.5, "materias": {"Programación III": 8.0, "Base de Datos II": 8.0, "Inglés Básico": 8.0, "Diseño de Circuitos Digitales": 8.0, "Estadística Computacional": 8.0}},
    {"nombres": "Diego Misael", "apellidos": "Villafranco Pineda", "edad": 20, "traslado": 30, "telefono": 8.0, "materias": {"Programación III": 9.0, "Base de Datos II": 9.0, "Inglés Básico": 9.0, "Diseño de Circuitos Digitales": 9.0, "Estadística Computacional": 9.0}},
    {"nombres": "Yeimy Julissa", "apellidos": "Parada Chávez", "edad": 20, "traslado": 45, "telefono": 3.0, "materias": {"Programación III": 9.0, "Base de Datos II": 9.0, "Inglés Básico": 9.0, "Diseño de Circuitos Digitales": 9.0, "Estadística Computacional": 9.0}},
    {"nombres": "Miguel Alejandro", "apellidos": "Gómez Saravia", "edad": 19, "traslado": 15, "telefono": 4.0, "materias": {"Programación III": 10.0, "Base de Datos II": 9.0, "Inglés Básico": 9.0, "Diseño de Circuitos Digitales": 9.0, "Estadística Computacional": 9.0}},
    {"nombres": "Josué Stiven", "apellidos": "Funes Benavides", "edad": 19, "traslado": 30, "telefono": 5.0, "materias": {"Programación III": 10.0, "Base de Datos II": 9.0, "Inglés Básico": 9.0, "Diseño de Circuitos Digitales": 9.0, "Estadística Computacional": 9.0}},
    {"nombres": "Jefferson Stiven", "apellidos": "Castillo Leiva", "edad": 19, "traslado": 15, "telefono": 2.0, "materias": {"Programación III": 10.0, "Base de Datos II": 9.0, "Inglés Básico": 9.0, "Diseño de Circuitos Digitales": 9.0, "Estadística Computacional": 9.0}},
    {"nombres": "Ramael Ernesto", "apellidos": "Cordova Colindres", "edad": 22, "traslado": 7, "telefono": 13.52, "materias": {"Programación III": 10.0, "Base de Datos II": 9.0, "Inglés Básico": 9.0, "Diseño de Circuitos Digitales": 9.0, "Estadística Computacional": 9.0}},
    {"nombres": "Carlos Daniel", "apellidos": "Parada Avalos", "edad": 25, "traslado": 10, "telefono": 5.52, "materias": {"Programación III": 9.0, "Base de Datos II": 9.0, "Inglés Básico": 9.0, "Diseño de Circuitos Digitales": 9.0, "Estadística Computacional": 9.0}},
    {"nombres": "Carlos David", "apellidos": "Turcios Rivas", "edad": 20, "traslado": 25, "telefono": 5.0, "materias": {"Programación III": 9.0, "Base de Datos II": 10.0, "Inglés Básico": 10.0, "Diseño de Circuitos Digitales": 9.0, "Estadística Computacional": 9.0}},
    {"nombres": "Gilda Mileidy", "apellidos": "González Castro", "edad": 19, "traslado": 30, "telefono": 5.92, "materias": {"Programación III": 10.0, "Base de Datos II": 10.0, "Inglés Básico": 10.0, "Diseño de Circuitos Digitales": 10.0, "Estadística Computacional": 10.0}},
    {"nombres": "Samuel Alexander", "apellidos": "Portillo Cerón", "edad": 20, "traslado": 80, "telefono": 6.0, "materias": {"Programación III": 10.0, "Base de Datos II": 10.0, "Inglés Básico": 10.0, "Diseño de Circuitos Digitales": 10.0, "Estadística Computacional": 10.0}},
    {"nombres": "Saúl Alexander", "apellidos": "Jimenez Garcia", "edad": 21, "traslado": 60, "telefono": 5.0, "materias": {"Programación III": 10.0, "Base de Datos II": 9.0, "Inglés Básico": 9.0, "Diseño de Circuitos Digitales": 9.0, "Estadística Computacional": 9.0}},
    {"nombres": "Edgar Ernesto", "apellidos": "Ramirez Barrera", "edad": 21, "traslado": 60, "telefono": 7.0, "materias": {"Programación III": 9.0, "Base de Datos II": 10.0, "Inglés Básico": 10.0, "Diseño de Circuitos Digitales": 10.0, "Estadística Computacional": 10.0}},
    {"nombres": "Jose Luis", "apellidos": "Garcia Mejia", "edad": 20, "traslado": 90, "telefono": 8.0, "materias": {"Programación III": 9.0, "Base de Datos II": 9.0, "Inglés Básico": 9.0, "Diseño de Circuitos Digitales": 8.0, "Estadística Computacional": 8.0}},
    {"nombres": "Jaime Shafik", "apellidos": "Salmeron Chicas", "edad": 20, "traslado": 7, "telefono": 6.0, "materias": {"Programación III": 8.0, "Base de Datos II": 10.0, "Inglés Básico": 10.0, "Diseño de Circuitos Digitales": 10.0, "Estadística Computacional": 8.0}},
    {"nombres": "Wilfredo Josue", "apellidos": "Ortez Gaitan", "edad": 20, "traslado": 30, "telefono": 5.0, "materias": {"Programación III": 9.0, "Base de Datos II": 10.0, "Inglés Básico": 10.0, "Diseño de Circuitos Digitales": 10.0, "Estadística Computacional": 8.0}},
    {"nombres": "Andrea María", "apellidos": "López Rivas", "edad": 19, "traslado": 20, "telefono": 4.5, "materias": {"Programación III": 9.5, "Base de Datos II": 9.0, "Inglés Básico": 10.0, "Diseño de Circuitos Digitales": 9.0, "Estadística Computacional": 9.5}}
]

class EstadisticaEstudiantil:
    def __init__(self, datos):
        self.datos = datos

    def media(self, lista):
        return sum(lista) / len(lista)

    def mediana(self, lista):
        ordenados = sorted(lista)
        n = len(lista)
        mitad = n // 2
        return (ordenados[mitad - 1] + ordenados[mitad]) / 2 if n % 2 == 0 else ordenados[mitad]

    def moda(self, lista):
        conteo = Counter(lista)
        max_veces = max(conteo.values())
        if max_veces == 1:
            return "No hay moda"
        return [f"{num} (se repite {veces} veces)" for num, veces in conteo.items() if veces == max_veces]

    def varianza(self, lista):
        m = self.media(lista)
        return sum((x - m) ** 2 for x in lista) / (len(lista) - 1)

    def desviacion_estandar(self, lista):
        return math.sqrt(self.varianza(lista))

    def desviacion_tipica(self, lista):
        return self.desviacion_estandar(lista)

    def rango(self, lista):
        return max(lista) - min(lista)

    def frecuencia(self, lista):
        return dict(Counter(lista))

    def obtener_valores_y_mostrar(self, campo):
        valores = []
        print("\n--- DATOS ANALIZADOS ---")
        if campo == "nota":
            for est in self.datos:
                promedio = sum(est["materias"].values()) / len(est["materias"])
                valores.append(promedio)
                print(f"{est['nombres']} {est['apellidos']} -> Promedio General: {promedio:.2f} pts")
        elif campo == "traslado":
            for est in self.datos:
                val = est[campo]
                valores.append(val)
                print(f"{est['nombres']} {est['apellidos']} -> Traslado: {val} min")
        elif campo == "telefono":
            for est in self.datos:
                val = est[campo]
                valores.append(val)
                print(f"{est['nombres']} {est['apellidos']} -> Teléfono: {val} hrs")
        elif campo == "edad":
            for est in self.datos:
                val = est[campo]
                valores.append(val)
                print(f"{est['nombres']} {est['apellidos']} -> Edad: {val} años")
        return valores

    def elegir_campo(self):
        print("\n¿De cuál dato quieres calcular la estadística?")
        print("1. Notas (Promedio de materias)")
        print("2. Edades")
        print("3. Tiempo de traslado (minutos)")
        print("4. Horas de uso de teléfono")
        op = input("Elige una opción: ")
        opciones = {"1": "nota", "2": "edad", "3": "traslado", "4": "telefono"}
        return opciones.get(op, None)

    def ejecutar_menu(self):
        while True:
            print("\n==================================")
            print("   SISTEMA ESTADÍSTICO ACADÉMICO  ")
            print("==================================")
            print("1. Media aritmética")
            print("2. Mediana")
            print("3. Moda")
            print("4. Varianza")
            print("5. Desviación estándar")
            print("6. Desviación típica")
            print("7. Rango")
            print("8. Frecuencia")
            print("9. Ver explicación teórica (Diferencias)")
            print("0. Salir")

            opcion = input("Elige una opción: ")

            if opcion == "0":
                print("¡Hasta luego!")
                break
            elif opcion == "9":
                self.mostrar_explicacion_teorica()
                input("\nPresione Enter para volver al menú...")
                continue

            if opcion in [str(i) for i in range(1, 9)]:
                campo = self.elegir_campo()
                if campo is None:
                    print("Opción de campo no válida.")
                    continue

                valores = self.obtener_valores_y_mostrar(campo)

                print("\n==================================")
                print("       RESULTADO ESTADÍSTICO      ")
                print("==================================")
                if opcion == "1":
                    print(f"Media: {self.media(valores):.2f}")
                elif opcion == "2":
                    print(f"Mediana: {self.mediana(valores):.2f}")
                elif opcion == "3":
                    print(f"Moda: {self.moda(valores)}")
                elif opcion == "4":
                    print(f"Varianza: {self.varianza(valores):.2f}")
                elif opcion == "5":
                    print(f"Desviación estándar: {self.desviacion_estandar(valores):.2f}")
                elif opcion == "6":
                    print(f"Desviación típica: {self.desviacion_tipica(valores):.2f}")
                elif opcion == "7":
                    print(f"Rango: {self.rango(valores):.2f}")
                elif opcion == "8":
                    print(f"Frecuencia: {self.frecuencia(valores)}")

                input("\nPresione Enter para volver al menú...")
            else:
                print("Opción no válida. Intente de nuevo.")

    def mostrar_explicacion_teorica(self):
        print("\n-----------------------------------------------------------")
        print(" EXPLICACIÓN CON NUESTRAS PALABRAS DE LAS DIFERENCIAS ")
        print("-----------------------------------------------------------")
        print("1. Desviación Estándar vs. Desviación Típica:")
        print("   En estadística son exactamente lo mismo; ambos términos se usan ")
        print("   como sinónimos para referirse a la raíz cuadrada de la varianza.")
        print("\n2. Varianza vs. Desviación Estándar/Típica:")
        print("   - La Varianza mide qué tan dispersos están los datos elevando ")
        print("     las diferencias al cuadrado (ej. minutos al cuadrado).")
        print("   - La Desviación Estándar/Típica le saca raíz cuadrada a la varianza ")
        print("     para expresar la dispersión en las mismas unidades originales ")
        print("     (ej. minutos exactos o notas).")
        print("-----------------------------------------------------------\n")

if __name__ == "__main__":
    app = EstadisticaEstudiantil(estudiantes)
    app.ejecutar_menu()
```
